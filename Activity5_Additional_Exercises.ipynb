## 1. Emergency Dispatch Record Search (rotated sorted list)

```python
print("=" * 60)
print("EXERCISE 1: EMERGENCY DISPATCH RECORD SEARCH")
print("=" * 60)

N = 20
STEP = SEED_DIGIT + 2
original = [ID_SUM + STEP * i for i in range(N)]          # unique, ascending
rotation = SEED_DIGIT % N                                  # rotation point
rotated = original[rotation:] + original[:rotation]
target = original[(ID_SUM + ARTIST_SUM) % N]               # target from ID_SUM and ARTIST_SUM

def linear_search(data, target):
    comparisons = 0
    for i, v in enumerate(data):
        comparisons += 1
        if v == target:
            return i, comparisons
    return -1, comparisons

def rotated_binary_search(data, target):
    """Binary search on a rotated sorted list. O(log n) time, O(1) space."""
    low, high, comparisons = 0, len(data) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons
        comparisons += 1
        if data[low] <= data[mid]:                 # left half is sorted
            comparisons += 1
            if data[low] <= target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:                                      # right half is sorted
            comparisons += 1
            if data[mid] < target <= data[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1, comparisons

def report(t):
    bi, bc = rotated_binary_search(rotated, t)
    li, lc = linear_search(rotated, t)
    print(f"Target incident: {t}")
    print("  Rotated binary search:", f"position {bi}" if bi >= 0 else "NOT FOUND", f"| comparisons = {bc}")
    print("  Linear search        :", f"position {li}" if li >= 0 else "NOT FOUND", f"| comparisons = {lc}")
    return bi == li

print("Original dataset:", original)
print("Rotation point  :", rotation)
print("Rotated dataset :", rotated)
print()
ok1 = report(target)
print()
print("Extra check (another existing value):")
ok2 = report(original[N // 2])
print("Extra check (value known to be missing):")
ok3 = report(original[-1] + 1)
print()
print("Results agree with linear search:", ok1 and ok2 and ok3)
print("Time complexity : O(log n) for rotated binary search vs O(n) for linear search")
print("Space complexity: O(1) (a few index variables only)")

print()
print("FINAL OUTPUT: target", target, "->", "found" if rotated_binary_search(rotated, target)[0] >= 0 else "not found")
```

## 2. University Enrollment Records (merge two sorted lists)

```python
print("=" * 60)
print("EXERCISE 2: UNIVERSITY ENROLLMENT RECORDS")
print("=" * 60)

dept_a = [ID_SUM + (SEED_DIGIT + 1) * i for i in range(8)]       # SEED_DIGIT, ID_SUM
dept_b = [NAME_LENGTH + ARTIST_LENGTH * i for i in range(10)]    # ARTIST_LENGTH, NAME_LENGTH

def merge_sorted(a, b):
    """Merge two sorted lists. O(m + n) time, O(m + n) space."""
    i = j = comparisons = 0
    merged = []
    while i < len(a) and j < len(b):
        comparisons += 1
        if a[i] <= b[j]:
            merged.append(a[i]); i += 1
        else:
            merged.append(b[j]); j += 1
    merged.extend(a[i:])
    merged.extend(b[j:])
    return merged, comparisons

merged, comps = merge_sorted(dept_a, dept_b)

print("Department A dataset:", dept_a)
print("Department B dataset:", dept_b)
print("Consolidated dataset:", merged)
print(f"Records before: {len(dept_a)} + {len(dept_b)} = {len(dept_a) + len(dept_b)} | after: {len(merged)}")
print("Comparison count:", comps)
print("Ordering verification:", merged == sorted(dept_a + dept_b) and all(x <= y for x, y in zip(merged, merged[1:])))
print("Time complexity : O(m + n)")
print("Space complexity: O(m + n)")

print()
print("FINAL OUTPUT: consolidated enrollment list =", merged)
```

## 3. E-Commerce Order Processing (sorting under different input orders)

```python
print("=" * 60)
print("EXERCISE 3: E-COMMERCE ORDER PROCESSING")
print("=" * 60)
import random, time

N_ORDERS = 100 + ID_SUM * 5
A = (ID_SUM * 7 + ARTIST_SUM) % 996 + 1                     # 1..996, coprime to 997
B = (SEED_DIGIT * 31 + NAME_LENGTH * 17 + ARTIST_LENGTH) % 997
orders = [(A * i + B) % 997 for i in range(N_ORDERS)]       # unique order numbers

sorted_input = sorted(orders)
reverse_input = sorted(orders, reverse=True)
random_input = orders[:]
random.Random(ID_SUM * 100 + SEED_DIGIT).shuffle(random_input)

def bubble_sort(data):
    a, comparisons, moves = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                moves += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, moves

def selection_sort(data):
    a, comparisons, moves = data.copy(), 0, 0
    for i in range(len(a) - 1):
        m = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[m]:
                m = j
        if m != i:
            a[i], a[m] = a[m], a[i]
            moves += 1
    return a, comparisons, moves

def insertion_sort(data):
    a, comparisons, moves = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            moves += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, moves

def timed(fn, data, trials=5):
    total = 0
    for _ in range(trials):
        t = time.perf_counter()
        result = fn(data)
        total += time.perf_counter() - t
    return result, total / trials

algorithms = [("Bubble Sort", bubble_sort), ("Selection Sort", selection_sort), ("Insertion Sort", insertion_sort)]
conditions = [("Sorted", sorted_input), ("Reverse", reverse_input), ("Random", random_input)]

print("Number of orders:", N_ORDERS)
print("Generated orders (first 10):", orders[:10], "...")
print("Sorted input  (first 10):", sorted_input[:10], "...")
print("Reverse input (first 10):", reverse_input[:10], "...")
print("Random input  (first 10):", random_input[:10], "...")
print()
print(f"{'Condition':<10}{'Algorithm':<16}{'Comparisons':>12}{'Movements':>11}{'Time (s)':>13}  Sorted?")

table = {}
expected = sorted(orders)
for cname, data in conditions:
    for aname, fn in algorithms:
        (res, comps, moves), secs = timed(fn, data)
        ok = res == expected
        table[(cname, aname)] = (comps, moves, secs)
        print(f"{cname:<10}{aname:<16}{comps:>12}{moves:>11}{secs:>13.8f}  {ok}")

print()
print("Big O analysis (n =", N_ORDERS, ")")
print("  Bubble    : best O(n) (sorted, early exit) | average/worst O(n^2)")
print("  Selection : O(n^2) in every case; at most n-1 swaps")
print("  Insertion : best O(n) (sorted) | average/worst O(n^2)")
print()
print("Best/average/worst observations")
for aname, _ in algorithms:
    best = table[("Sorted", aname)][0]; avg = table[("Random", aname)][0]; worst = table[("Reverse", aname)][0]
    print(f"  {aname:<15} best(sorted)={best:>6} | average(random)={avg:>6} | worst(reverse)={worst:>6} comparisons")

print()
print("FINAL OUTPUT: all sorts verified correct on all three input conditions;")
print("  Bubble and Insertion Sort are fastest on sorted input and slowest on reverse input,")
print("  and Selection Sort does the same number of comparisons regardless of input order.")
```
